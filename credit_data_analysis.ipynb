# Credit Data Analysis with Python

This project explores a credit application dataset using Python to analyze patterns in debt, employment, demographics, and credit approval.

The analysis uses data visualization and correlation analysis to better understand relationships within the dataset.

## Tools Used
- Python
- Pandas
- NumPy
- Seaborn
- Matplotlib
- Google Colab

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
sns.set(style='whitegrid')
```

## Load the Dataset

The dataset is loaded into a Pandas DataFrame. The first five rows are displayed to preview the structure and variables in the data.

```python
df = pd.read_csv('clean_dataset.csv')
df.head()
```

## 1. Debt Distribution by Ethnicity

This analysis compares the distribution of debt across different ethnicity groups using a box plot.

```python
# Create a box plot to compare debt across ethnicity groups
sns.catplot(data=df, x='Ethnicity', y='Debt', kind='box',
            color='skyblue')
plt.title('Debt by Ethnicity', fontsize=14, fontweight='bold'
, color='black')
plt.xlabel('Ethnicity', fontsize=12)
plt.ylabel('Debt Amount', fontsize=12)
plt.grid(True)
plt.show()
```

### Interpretation

The box plot shows how debt is distributed across the different ethnicity
groups in the dataset.

The Other group has the highest median debt, while the White and Latino groups have lower median debt. The different box sizes and outliers also show that debt varies within each group.


## 2. Average Debt by Industry and Gender

This analysis compares the average debt of males and females across different industries.

```python
# Compare average debt by industry and gender

grouped_df = df.copy()
grouped_df['Gender'] = grouped_df['Gender'].replace({0: 'Female', 1: 'Male'})
grouped_df = grouped_df.groupby(['Industry', 'Gender'], as_index=False)['Debt'].mean()
plt.figure(figsize=(10, 5))
sns.pointplot(data=grouped_df, x='Industry', y='Debt', hue='Gender',
              palette='Set2', markers='*', markersize=10,
              markerfacecolor='pink',markeredgewidth=2, markeredgecolor='purple'
              , linestyles='-')
plt.title('Average Debt by Industry (Male vs Female)', fontsize=14,
          fontweight='bold', color='black')
plt.xlabel('Industry', fontsize=12, color='darkgreen')
plt.ylabel('Average Debt Amount', fontsize=12, color='darkgreen')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Interpretation

The chart compares the average debt of males and females across different industries.

Average debt varies by both industry and gender. For example, males have higher average debt in Healthcare, while females have higher average debt in Financials and Real Estate. In some industries, such as Education and Consumer Staples, the average debt is very similar for both groups.

Overall, the differences between males and females are larger in some industries and smaller in others.

## 3. Correlation Analysis

This analysis examines the relationships between debt, years employed, age, and credit approval using a correlation heatmap.

```python
# Create a correlation heatmap for selected numerical variables

plt.figure(figsize=(8, 6))
heatmap_data = df[['Debt', 'YearsEmployed', 'Age', 'Approved']]
sns.heatmap(heatmap_data.corr(), annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap: Debt, Years Employed, Age, Approved',
          fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### Interpretation

The heatmap shows the correlations between debt, years employed, age, and credit approval.

1. Age and years employed have the strongest positive correlation at 0.39.
2. Years employed and credit approval have a positive correlation of 0.32.
3. Debt and years employed have a positive correlation of 0.30.

The remaining correlations are also positive but relatively weak. Overall, none of the variables shown have a strong correlation with each other.

## 4. Years Employed vs. Debt

This analysis examines the relationship between years of employment and debt using a scatter plot with a regression line.

```python
# Create a scatter plot with a regression line for years employed and debt
plt.figure(figsize=(8, 6))
sns.regplot(data=df, x='YearsEmployed', y='Debt', color='skyblue', marker='o')
plt.title('Years Employed vs Debt', fontsize=14, fontweight='bold',
          color='black')
plt.xlabel('Years Employed', fontsize=12, color='darkgreen')
plt.ylabel('Debt Amount', fontsize=12, color='darkgreen')
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Interpretation

The scatter plot shows the relationship between years employed and debt. Each point represents an observation in the dataset.

The regression line slopes upward, showing a positive relationship between the two variables. This is consistent with the correlation of 0.30 found in the previous analysis.

However, the points are widely spread around the regression line, indicating that the relationship between years employed and debt is relatively weak.

The analysis examined debt across ethnicity groups, compared average debt across industries and gender, analyzed correlations between numerical variables, and explored the relationship between years employed and debt.

Overall, the analysis showed that debt varies across different demographic and employment groups. The correlation analysis also showed positive but relatively weak relationships among debt, age, years employed, and credit approval.
